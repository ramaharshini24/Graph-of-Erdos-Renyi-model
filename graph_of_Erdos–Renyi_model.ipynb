```python
# ================================================================
# PROBLEM 1
# Multi-Agent Formation Control
# Displaying the name HARSHINI using N = 20 agents
# ================================================================

import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation

from IPython.display import Video, display
from google.colab import files

# ================================================================
# 0. Reproducibility
# ================================================================

np.random.seed(7)


# ================================================================
# 1. Parameters
# ================================================================

N = 20

# Erdos-Renyi connection probability
P_ER = 0.35

# Name to display
NAME = "HARSHINI"

# Random initial position range
WORLD_SCALE = 10.0

# Letter dimensions
LETTER_WIDTH = 4.0
LETTER_HEIGHT = 6.0

# Scale of displayed letters
LETTER_GAIN = 1.4

# Formation-control simulation parameters
DT = 0.03

T_PER_LETTER = 4.0

HOLD_TIME = 0.8

STEPS_PER_LETTER = int(
    T_PER_LETTER / DT
)

HOLD_STEPS = int(
    HOLD_TIME / DT
)

# Video parameters
FPS = 30

FRAME_STRIDE = max(
    1,
    int(
        round(
            1 / (FPS * DT)
        )
    )
)


# ================================================================
# 2. Generate connected Erdos-Renyi graph
# ================================================================

def make_connected_er_graph(
    n,
    p,
    seed=None
):

    rng = np.random.RandomState(seed)

    trial = 0

    while True:

        trial += 1

        if seed is None:

            s = None

        else:

            s = rng.randint(
                0,
                10**6
            )

        G = nx.erdos_renyi_graph(
            n,
            p,
            seed=s
        )

        # Keep generating until connected
        if nx.is_connected(G):

            return G, trial


# Generate graph
G, n_trials = make_connected_er_graph(
    N,
    P_ER,
    seed=42
)

# Graph Laplacian
L = nx.laplacian_matrix(
    G
).toarray().astype(float)


print(
    f"Erdos-Renyi graph G({N}, {P_ER}) "
    f"generated and connected after "
    f"{n_trials} trial(s)."
)

print(
    f"Number of edges: "
    f"{G.number_of_edges()}"
)

print(
    f"Graph connected: "
    f"{nx.is_connected(G)}"
)


# ================================================================
# 3. Random initial positions in R^2
# ================================================================

x0 = (
    np.random.rand(N, 2)
    - 0.5
) * WORLD_SCALE


print("\nInitial agent positions:")
print(x0)


# ================================================================
# 4. Define capital letters
# ================================================================

def letter_segments(ch):

    w = LETTER_WIDTH
    h = LETTER_HEIGHT


    # ------------------------------------------------------------
    # H
    # ------------------------------------------------------------

    if ch == "H":

        return [
            (0, 0, 0, h),
            (w, 0, w, h),
            (0, h / 2, w, h / 2)
        ]


    # ------------------------------------------------------------
    # A
    # ------------------------------------------------------------

    if ch == "A":

        return [
            (0, 0, w / 2, h),
            (w / 2, h, w, 0),
            (
                w * 0.22,
                h * 0.42,
                w * 0.78,
                h * 0.42
            )
        ]


    # ------------------------------------------------------------
    # R
    # ------------------------------------------------------------

    if ch == "R":

        return [

            # Vertical stem
            (0, 0, 0, h),

            # Top horizontal
            (0, h, 0.75 * w, h),

            # Upper curved section
            (
                0.75 * w,
                h,
                w,
                0.8 * h
            ),

            # Middle curve
            (
                w,
                0.8 * h,
                0.75 * w,
                0.6 * h
            ),

            # Middle horizontal
            (
                0.75 * w,
                0.6 * h,
                0,
                0.6 * h
            ),

            # Diagonal leg
            (
                0.2 * w,
                0.6 * h,
                w,
                0
            )
        ]


    # ------------------------------------------------------------
    # S
    # ------------------------------------------------------------

    if ch == "S":

        p = [

            (w, h),
            (0, h),
            (0, 0.55 * h),
            (w, 0.55 * h),
            (w, 0),
            (0, 0)

        ]

        return [

            (
                p[i][0],
                p[i][1],
                p[i + 1][0],
                p[i + 1][1]
            )

            for i in range(
                len(p) - 1
            )
        ]


    # ------------------------------------------------------------
    # I
    # ------------------------------------------------------------

    if ch == "I":

        return [

            # Top
            (
                0.25 * w,
                h,
                0.75 * w,
                h
            ),

            # Vertical
            (
                0.5 * w,
                h,
                0.5 * w,
                0
            ),

            # Bottom
            (
                0.25 * w,
                0,
                0.75 * w,
                0
            )
        ]


    # ------------------------------------------------------------
    # N
    # ------------------------------------------------------------

    if ch == "N":

        return [

            # Left vertical
            (0, 0, 0, h),

            # Diagonal
            (0, h, w, 0),

            # Right vertical
            (w, 0, w, h)
        ]


    raise ValueError(
        f"Letter '{ch}' is not defined."
    )


# ================================================================
# 5. Sample exactly 20 points along letter
# ================================================================

def sample_points_on_segments(
    segments,
    n_points
):

    segs = np.array(
        segments,
        dtype=float
    )

    starts = segs[:, 0:2]

    ends = segs[:, 2:4]


    # Length of every stroke
    seg_lengths = np.linalg.norm(
        ends - starts,
        axis=1
    )

    seg_lengths = np.maximum(
        seg_lengths,
        1e-9
    )


    total_length = (
        seg_lengths.sum()
    )


    cumulative_length = np.concatenate(
        [
            [0],
            np.cumsum(
                seg_lengths
            )
        ]
    )


    # Evenly distributed points
    sample_distances = np.linspace(
        0,
        total_length,
        n_points,
        endpoint=True
    )


    points = np.zeros(
        (
            n_points,
            2
        )
    )


    for k, d in enumerate(
        sample_distances
    ):

        segment_index = (
            np.searchsorted(
                cumulative_length,
                d,
                side="right"
            )
            - 1
        )


        segment_index = min(
            segment_index,
            len(segments) - 1
        )


        local_distance = (
            d
            - cumulative_length[
                segment_index
            ]
        )


        fraction = (
            local_distance
            / seg_lengths[
                segment_index
            ]
        )


        points[k] = (
            starts[
                segment_index
            ]
            +
            fraction
            *
            (
                ends[
                    segment_index
                ]
                -
                starts[
                    segment_index
                ]
            )
        )


    return points


# ================================================================
# 6. Generate target positions for every letter
# ================================================================

def letter_targets(
    ch,
    n_points=N
):

    points = sample_points_on_segments(
        letter_segments(ch),
        n_points
    )


    # Center letter around origin
    points = (
        points
        -
        np.array(
            [
                LETTER_WIDTH / 2,
                LETTER_HEIGHT / 2
            ]
        )
    )


    # Scale letter
    points *= LETTER_GAIN


    return points


# ================================================================
# 7. Generate HARSHINI sequence
# ================================================================

letter_sequence = list(
    NAME
)

print(
    "\nFormation sequence:"
)

print(
    " -> ".join(
        letter_sequence
    )
)


# Generate 20 target points for every letter
targets = [

    letter_targets(
        ch
    )

    for ch in letter_sequence

]


# ================================================================
# 8. Match agents between consecutive formations
# ================================================================

def greedy_assignment(
    src_pts,
    dst_pts
):

    src = src_pts.copy()

    dst = dst_pts.copy()

    n = src.shape[0]


    assigned = -np.ones(
        n,
        dtype=int
    )

    used = np.zeros(
        n,
        dtype=bool
    )


    for i in range(n):

        distances = np.linalg.norm(
            dst[~used]
            - src[i],
            axis=1
        )


        available_indices = (
            np.where(
                ~used
            )[0]
        )


        j = available_indices[
            np.argmin(
                distances
            )
        ]


        assigned[i] = j

        used[j] = True


    return dst[
        assigned
    ]


# Match initial random positions
# to the first letter
targets[0] = greedy_assignment(
    x0,
    targets[0]
)


# Match each letter to the next
for k in range(
    1,
    len(targets)
):

    targets[k] = greedy_assignment(
        targets[k - 1],
        targets[k]
    )


# ================================================================
# 9. Formation control
# ================================================================

def simulate_to_target(
    x_start,
    x_target,
    L,
    steps,
    dt
):

    """
    Formation control law:

        x_dot = -L(x - x_target)

    where L is the graph Laplacian.
    """


    trajectory = np.zeros(
        (
            steps + 1,
            x_start.shape[0],
            2
        )
    )


    x = x_start.copy()

    trajectory[0] = x


    for t in range(steps):

        # Formation-control input
        u = -L @ (
            x - x_target
        )


        # Euler integration
        x = (
            x
            + dt * u
        )


        trajectory[
            t + 1
        ] = x


    return trajectory


# ================================================================
# 10. Generate all animation frames
# ================================================================

all_frames = [
    x0.copy()
]

letter_labels = [
    "START"
]


x_current = x0.copy()


for k, ch in enumerate(
    letter_sequence
):

    print(
        f"Generating formation: {ch}"
    )


    # Move to target letter
    trajectory = simulate_to_target(
        x_current,
        targets[k],
        L,
        STEPS_PER_LETTER,
        DT
    )


    # ------------------------------------------------------------
    # Movement frames
    # ------------------------------------------------------------

    for t in range(
        1,
        trajectory.shape[0],
        FRAME_STRIDE
    ):

        all_frames.append(
            trajectory[t]
        )

        letter_labels.append(
            ch
        )


    # ------------------------------------------------------------
    # Hold formed letter
    # ------------------------------------------------------------

    for _ in range(
        0,
        HOLD_STEPS,
        FRAME_STRIDE
    ):

        all_frames.append(
            trajectory[-1]
        )

        letter_labels.append(
            ch
        )


    x_current = (
        trajectory[-1]
    )


all_frames = np.array(
    all_frames
)


print(
    "\nTotal frames:",
    all_frames.shape[0]
)

print(
    "Approximate video duration:",
    f"{all_frames.shape[0] / FPS:.2f} seconds"
)


# ================================================================
# 11. Create figure
# ================================================================

fig, (
    ax_main,
    ax_graph
) = plt.subplots(
    1,
    2,
    figsize=(14, 7)
)


# ================================================================
# 12. Plot limits
# ================================================================

pad = 3

xmin = (
    all_frames[:, :, 0].min()
    - pad
)

xmax = (
    all_frames[:, :, 0].max()
    + pad
)

ymin = (
    all_frames[:, :, 1].min()
    - pad
)

ymax = (
    all_frames[:, :, 1].max()
    + pad
)


# ================================================================
# 13. Main formation-control panel
# ================================================================

ax_main.set_xlim(
    xmin,
    xmax
)

ax_main.set_ylim(
    ymin,
    ymax
)

ax_main.set_aspect(
    "equal"
)

ax_main.set_title(
    "Formation Control: HARSHINI",
    fontsize=16,
    fontweight="bold"
)

ax_main.set_xlabel(
    "x"
)

ax_main.set_ylabel(
    "y"
)


# ================================================================
# 14. Communication edges
# ================================================================

edge_lines = [

    ax_main.plot(
        [],
        [],
        color="lightsteelblue",
        linewidth=1.2,
        zorder=1
    )[0]

    for _ in G.edges()

]


# ================================================================
# 15. AGENTS
#     ALL 20 AGENTS ARE RED
# ================================================================

scat = ax_main.scatter(

    x0[:, 0],
    x0[:, 1],

    s=90,

    c="red",

    zorder=3,

    edgecolors="black",

    linewidths=0.7
)


# ================================================================
# 16. Current letter label
# ================================================================

letter_text = ax_main.text(

    0.5,
    1.02,

    "",

    transform=ax_main.transAxes,

    ha="center",

    va="bottom",

    fontsize=28,

    fontweight="bold"
)


# ================================================================
# 17. Right-side communication graph
# ================================================================

pos_graph = nx.spring_layout(
    G,
    seed=1
)


nx.draw(

    G,

    pos_graph,

    ax=ax_graph,

    node_color="red",

    node_size=600,

    edgecolors="black",

    linewidths=0.7,

    with_labels=True,

    font_size=9,

    font_color="white"
)


ax_graph.set_title(

    f"Erdos-Renyi Communication Graph\n"
    f"G({N}, p={P_ER}) | "
    f"{G.number_of_edges()} edges",

    fontsize=14,

    fontweight="bold"
)


# ================================================================
# 18. Animation update function
# ================================================================

edges = list(
    G.edges()
)


def update(frame_idx):

    # Current positions
    points = all_frames[
        frame_idx
    ]


    # Update agent positions
    scat.set_offsets(
        points
    )


    # Update communication edges
    for line, (i, j) in zip(
        edge_lines,
        edges
    ):

        line.set_data(

            [
                points[i, 0],
                points[j, 0]
            ],

            [
                points[i, 1],
                points[j, 1]
            ]

        )


    # Current letter
    current_label = (
        letter_labels[
            frame_idx
        ]
    )


    if current_label != "START":

        letter_text.set_text(
            current_label
        )

    else:

        letter_text.set_text(
            ""
        )


    return (
        [scat, letter_text]
        + edge_lines
    )


# ================================================================
# 19. Create animation
# ================================================================

ani = animation.FuncAnimation(

    fig,

    update,

    frames=all_frames.shape[0],

    interval=1000 / FPS,

    blit=False
)


# ================================================================
# 20. Save MP4
# ================================================================

output_video = (
    "HARSHINI_formation_control.mp4"
)


print(
    "\nSaving MP4..."
)


writer = animation.FFMpegWriter(

    fps=FPS,

    bitrate=3000
)


ani.save(

    output_video,

    writer=writer,

    dpi=120
)


plt.close(fig)


print(
    f"Video saved successfully: "
    f"{output_video}"
)


# ================================================================
# 21. DISPLAY VIDEO INSIDE COLAB OUTPUT CELL
# ================================================================

print(
    "\nDisplaying video:"
)


display(
    Video(
        output_video,
        embed=True
    )
)


# ================================================================
# 22. DOWNLOAD VIDEO
# ================================================================


files.download(
    output_video
)
```
